```python
d=[2,14,18,7,12,9,20,11,3,7,15,10]

# dict={'elev_1':2,'elev_2':14,'elev_3':18,'elev_4':7}
# print(dict)

for idx, i in enumerate(d,start=1):
    if i<8:
        cat='mauvais'
    elif i<=14:
        cat="moyen"
    else:
        cat="bon"
    print(f"Note de l'élève n°{idx} : {i} Catégorie : {cat}")

def moyenne(liste):
    return sum(liste)/len(liste)
moyenne(d)
```

# RAPPORT 

##### Téléchargement des librairies utilisées 

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import numpy as np 
from sklearn.datasets import load_breast_cancer

```

##### Téléchargement du dataset 

###### Utilisation du dataset contenant des données de patient ayant contracté le cancer du sein dans le Wisconsin 

```python
df = load_breast_cancer(as_frame=True).frame
```

###### Visualisation des premières données du df 

```python
df.head()
```

##### Vérifications des colonnes pour n'en garder qu'un certain nombre 

```python
df.columns
```

###### Beaucoup de trop de colonnes, on ne va en garder que quelques unes pour réguler la mémoire utilisée 

```python
col=['mean radius', 'mean texture', 'mean perimeter', 'mean area',
       'mean smoothness', 'mean compactness', 'mean concavity','worst radius', 'worst texture', 'worst perimeter', 'worst area',
       'worst smoothness', 'worst compactness', 'worst concavity','target']
df=df[col]
df
```

##### types des colonnes à vérifier pour traiter les données 

```python
df.dtypes
```

###### note : que du float donc des données facilement traitables

```python
df.isna()
df.isna().sum()
```

###### note : pas de valeurs manquantes donc on peut commencer à traiter et exploitier les données 

##### Visualisation des distributions 

###### par exemple, sur l'aire moyenne de la tumeur 

```python
# utilisation de matplotlib 

plt.figure(figsize=(3,4))
plt.hist(df['mean area'],bins=15,color='skyblue',edgecolor='black')
plt.ylabel("nombre d'échantillons")
plt.xlabel('Aire moyenne')
plt.title("Histogramme de l'aire moyenne des cancers de la population étudiée")
plt.show()
```

###### note : la majorité des tumeurs ont une aire concentrée autour de 400 unité puis on observe que le fait d'avoir une tumeur plus imposante est davantage rare 

```python
sns.histplot(df['mean radius'],color='red',kde=True)
plt.title('Mean Radius')
plt.show()
```

###### note : Observation similaire pour la variable mean radius 

##### Nuage de point sur une autre variable 

```python
sns.scatterplot(x=df['mean radius'],y=df['mean texture'],s=100,hue=df['target'])
plt.title('Aire des sépales par espèce')
plt.show()
```

###### note : on peut y rajouter un paramètre afin de classifier les points selon le variable target pour obtenir davantage d'information 

###### note2 : on voit que les deux classes de target sont très distinctes l'un de l'autre, donc les informations du dataset sont très informatives et significatives pour expliquer la variable cible 

###### note3 : on peut produire une matrice de corrélation pour voir quelles sont les variables les plus corréelées entre elles 

```python
corr_matrice=df.corr()
plt.figure(figsize=(10,10))
sns.heatmap(corr_matrice, annot=True,fmt=".2f",cmap="coolwarm",cbar=True)
plt.title("matrice de corrélation des variables du cancer du sein")
plt.show()
```

###### note : beaucoup de corrélation positive ce qui expliquerait que lorsque la valeur d'une variable augmente, celle des autres augmente aussi. Point d'interrogation sur la variable cible qui a une corrélation négative avec la totalité des variables. Des méthodes de traitement devront être mis en places pour observer et minimser l'existence d'un biais.
