```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

### Premier test de récup d'un premier site

```python
url="https://books.toscrape.com/catalogue/category/books/travel_2/index.html"
```

```python
response=requests.get(url)

print(response.status_code)
html=response.text
print(html[:1000])
```

###### Note : la réponse 200 signifie le succès de la requête 

### Extraction des données souhaitées 

```python
soup = BeautifulSoup(html,'html.parser')

titres=soup.find_all("a")
for el in titres:
    if el.has_attr('title'):
        print(el.text.strip())

prix=soup.find_all("p", {"class" : "price_color"})
for el in prix:
    print(el.text.strip())

stock=soup.find_all('p',{"class" : "instock availability"})
for el in stock:
    text=el.get_text(strip=True)
    if text == "In stock":
         print("Yes")
    else:
        print('Not in stock')
```

### Prochaine étape : Rendre ça plus propre et les mettre en dataframe 

```python
data=[]

livres = soup.find_all('article', {'class':'product_pod'})
for livre in livres:
    titres=livre.h3.a['title']
    prix=livre.find('p', {'class':'price_color'}).text.strip()
    stock=livre.find('p',{ 'class':'instock availability'}).get_text(strip=True)
    data.append({
        "titre" : titres,
        "prix" : prix,
        "stock" : stock
    })
    print(titres,prix,stock)
```

```python
df= pd.DataFrame(data)
print(df)
```
